```python
import pandas as pd
df = pd.read_csv("bonus.csv")
df.head()
```

```python
from sklearn.model_selection import train_test_split

X = df[['performance', 'years_of_experience', 'projects_completed']].values
y = df[['bonus']].values 
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

### Train The Neural Network

```python
import torch
import torch.nn as nn
import torch.optim as optim

torch.manual_seed(0) 
#This ensures the model's initial weights are the same every time you run the code. 

X_train_tensor = torch.tensor(X_train,dtype = torch.float32)
y_train_tensor = torch.tensor(y_train,dtype = torch.float32)
X_test_tensor = torch.tensor(X_test, dtype=torch.float32)
y_test_tensor = torch.tensor(y_test, dtype=torch.float32)
#float32 uses half the memory and is faster on GPUs.
```

```python
class BonusPredictor(nn.Module):
    def __init__(self):
        super().__init__()
        self.network = nn.Sequential(
            nn.Linear(3, 1),  #single-layer network (one trainable layer).
        )

    def forward(self, x): #The forward() method defines how the input passes through the network.
        return self.network(x)
```

```python
model = BonusPredictor()
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters() , lr=0.005)
```

```python
# Training loop
epochs = 5000
for epoch in range(epochs):
    #Compute predicted bonus by passing X_train
    predictions = model(X_train_tensor)
    loss = criterion(predictions, y_train_tensor)

    #  Compute gradients
    optimizer.zero_grad()
    loss.backward()

    # Update parameters
    optimizer.step()

    # Print loss every 100 epochs
    if (epoch + 1) % 100 == 0:
        print(f"Epoch [{epoch + 1}/{epochs}], Loss: {loss.item():.2f}")
```

```python
# Switch to evaluation mode 
model.eval()

# Disable gradient calculations for evaluation to save memory
with torch.no_grad():
    test_predictions = model(X_test_tensor)
    test_loss = criterion(test_predictions, y_test_tensor)

print(f"Test Loss: {test_loss.item():.4f}")
```
